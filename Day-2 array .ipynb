```python
#  Find the largest number in a list of numbers
arr =[3,78,90,23,112,67]
larg = arr[0]
for i in arr:
    if i> larg:
        larg = i
print("The largest number in the list is:", larg)
```

```python
# Find the second largest number in a list of numbers
arr = [45,67,2,3,87,12]
larg = arr[0]
second_larg = arr[0]
for i in arr:
    if i > larg:
        second_larg = larg
        larg = i

print("The second largest number in the list is:", second_larg)
```

```python
# Count occurrences
arr = [1,2,3,4,5,1,2,3,1]
duplicates = {}
for i in arr:
    if i in duplicates:
        duplicates[i] += 1
    else:
        duplicates[i] = 1

print("Occurrences of each element:", duplicates)
```

```python
# Move all zeroes to the end
arr =[1,0,2,0,3,4,0,5]
count = 0
for i in range(len(arr)):
    if arr[i] != 0:
        arr[count] = arr[i]
        count+=1

# Fill the remaining positions with zeroes
for i in range(count, len(arr)):
    arr[i] = 0

print("Array after moving zeroes to the end:", arr)
```
