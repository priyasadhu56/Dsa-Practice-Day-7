```python
# Print all elements
arr = [1,3,8,56,43]
for i in arr:
    print(i)
```

```python
# Find the maximum
arr = [3,7,8,9,5,7,23,2]
max_value = arr[0]
for i in arr:
    if i > max_value:
        max_value = i
print("Maximum value is:", max_value)
```

```python
# Find the minimum
arr = [3,5,90,65,1,6]
min_value = arr[0]
for i in arr:
    if i< min_value:
        min_value = i
print("Minimum value is:", min_value)
```

```python
# Calculate the sum
arr = [6,8,9,34,23]
sum = 0
for i in arr:
    sum += i
print("Sum of all elements is:", sum)    
```

```python
# Count even and odd numbers
arr =[3,6,7,8,2,20,5]
even_count = 0
odd_count = 0
for i in arr:
    if i % 2 == 0:
        even_count += 1
else:
        odd_count +=1
print("Even numbers count:", even_count)        
print("Odd numbers count:", odd_count)        
```

```python
# Linear search
arr = [5,7,8,9,3,2,8]
search = 3
for i in arr:
    if i == search:
        print("Element found:", search)
        
```
