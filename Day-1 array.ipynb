```python
# Find the maximum
arr = [3,7,8,9,5,7,23,2]
max_value = arr[0]
for i in arr:
    if i > max_value:
        max_value = i
print("Maximum value is:", max_value)
```

```python
# Find the minimum
arr = [3,5,90,65,1,6]
min_value = arr[0]
for i in arr:
    if i< min_value:
        min_value = i
print("Minimum value is:", min_value)
```
