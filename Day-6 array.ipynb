```python
#  Find the first element
arr = [1, 2, 3, 4, 5]
first_element = arr[0]
for i in arr:
    if i == first_element:
        print("The first element is:", i)
```

```python
# Print the last element
arr = [1, 2, 3, 4, 5]
last_element = arr[-1]
for i in arr:
    if i == last_element:
        print("The last element is:", i)
```

```python
#  Print elements in reverse order
arr = [1, 2, 3, 4, 5]
for i in range(len(arr)-1, -1, -1): 
    print(arr[i])
```
