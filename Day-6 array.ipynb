```python
#  Find the first element
arr = [1, 2, 3, 4, 5]
first_element = arr[0]
for i in arr:
    if i == first_element:
        print("The first element is:", i)
```

```python
# Print the last element
arr = [1, 2, 3, 4, 5]
last_element = arr[-1]
for i in arr:
    if i == last_element:
        print("The last element is:", i)
```

```python
#  Print elements in reverse order
arr = [1, 2, 3, 4, 5]
for i in range(len(arr)-1, -1, -1): 
    print(arr[i])
```

```python
# Find the sum of even numbers
arr = [9,7,5,6,2,67,4,3,2,1]
even_sum = 0
for i in arr:
    if i % 2 == 0:
        even_sum += i
print("The sum of even numbers is:", even_sum)
```

```python
# Find the sum of odd numbers
arr = [9,7,5,6,2,67,4,3,2,1]
odd_sum = 0
for i in arr:
    if i % 2 != 0:
        odd_sum += i
print("The sum of odd numbers is:", odd_sum)
```

```python
# Find the largest even number

arr = [9,7,5,6,2,67,4,3,2,1]
largest_even = 0
for i in arr:
    if i % 2 == 0 and i > largest_even:
        largest_even = i
print("The largest even number is:", largest_even)
```
