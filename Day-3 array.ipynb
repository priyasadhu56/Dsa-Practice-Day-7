```python
# Move all zeroes to the end
arr =[8,0,9,0,4,3,0,9,8,0,1,0,2]
count = 0
for i in range(len(arr)):
    if arr[i] != 0:
        arr[count] = arr[i]
        count += 1
for i in range(count, len(arr)):
    arr[i] = 0
print(arr)
```

```python
#Remove duplicates from a sorted array
arr =[1,3,4,2,8,7,1,7,5,5]
duplicates = []
for i in arr:
    if i not in duplicates:
        duplicates.append(i)
print(duplicates)
        
```
